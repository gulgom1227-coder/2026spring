````python
# 예시: 배달 음식 비용 계산하기
pizza_price = 26000
num_pizza = 2

drink_price = 2000
num_drink = 3

delivery_fee = 3000
coupon = 4000
person_count = 3

# 1. 전체 주문 금액을 계산한다.
food_price = pizza_price*num_pizza+drink_price*num_drink
total_price = food_price+delivery_fee

# 2. 쿠폰 적용 후 결제 금액을 계산한다.
total_price = total_price-coupon # 값의 업데이트

# 3. 한 명당 금액을 계산한다.
print(total_price/person_count)
````
